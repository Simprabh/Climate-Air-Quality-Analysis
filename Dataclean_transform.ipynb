```python
# ==========================================
# STAGE 1: IMPORT LIBRARIES AND SETUP
# ==========================================

# pandas: data loading, cleaning and analysis
import pandas as pd

# numpy: numerical calculations
import numpy as np

# matplotlib and seaborn: data visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# pathlib: working with file and folder paths
from pathlib import Path

# Display settings for easier reading
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)

# Set a consistent style for our visualisations
sns.set_theme(style="whitegrid")

# Create an output folder for cleaned data and analysis results
output_folder = Path("output")
output_folder.mkdir(exist_ok=True)

print("Libraries imported successfully!")
print("Workspace is ready.")
```

```python
# ==========================================
# STAGE 2: LOAD AND INSPECT THE DATASET
# ==========================================

# Load the original CSV dataset
file_path = Path("../data/raw/delhi-weather-aqi-2025.csv")

df = pd.read_csv(file_path)

# Display the first five rows
print("First five rows:")
display(df.head())

# Check the number of rows and columns
print("\nDataset shape:")
print(df.shape)

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()

# Generate basic statistical summaries
print("\nNumerical summary:")
display(df.describe())

# Check the number of unique monitoring locations
print("\nUnique locations:")
print(df["location"].unique())

print("\nNumber of unique locations:", df["location"].nunique())
```

```python
# ==========================================
# STAGE 3: DATA QUALITY ASSESSMENT
# ==========================================

# 1. Check missing values in each column
print("Missing values:")
display(df.isnull().sum())

# 2. Calculate percentage of missing values
print("\nMissing value percentage:")
display((df.isnull().mean() * 100).round(2))

# 3. Check exact duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# 4. Check duplicate date-time-location records
print("\nDuplicate location-time records:")
print(df.duplicated(subset=["date_ist", "time_ist", "location"]).sum())

# 5. Check data types
print("\nColumn data types:")
display(df.dtypes)

# 6. Check dataset dimensions
print("\nTotal rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
# ==========================================
# STAGE 4: DATE AND TIME VALIDATION
# ==========================================

# Check the first and last date values
print("Date range in original dataset:")
print("First date:", df["date_ist"].min())
print("Last date:", df["date_ist"].max())

# Check unique date values
print("\nNumber of unique dates:")
print(df["date_ist"].nunique())

# Check unique time values
print("\nNumber of unique time values:")
print(df["time_ist"].nunique())

# Check the first few date and time records
print("\nSample date and time records:")
display(df[["date_ist", "time_ist", "location"]].head(10))

# Convert date column using day/month/year format
df["date_ist"] = pd.to_datetime(
    df["date_ist"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Check whether any dates failed to convert
print("\nInvalid dates after conversion:")
print(df["date_ist"].isna().sum())

# Check records per location
print("\nRecords per location:")
display(df.groupby("location").size())
```

```python
# ==========================================
# STAGE 5: CHECK HOURLY DATA COMPLETENESS
# ==========================================

# Count hourly records for each location and date
daily_record_counts = (
    df.groupby(["location", "date_ist"])
    .size()
    .reset_index(name="hourly_records")
)

# Check how many days have exactly 24 records
print("Hourly records per location-day:")
display(daily_record_counts["hourly_records"].value_counts().sort_index())

# Identify days with fewer or more than 24 records
incomplete_days = daily_record_counts[
    daily_record_counts["hourly_records"] != 24
]

print("\nNumber of incomplete location-days:")
print(len(incomplete_days))

# Display incomplete records if any exist
print("\nIncomplete location-days:")
display(incomplete_days.head(20))

# Final validation
if len(incomplete_days) == 0:
    print("SUCCESS: Every location has exactly 24 hourly records per day.")
else:
    print("WARNING: Some location-days do not have 24 hourly records.")
```

```python
# ==========================================
# STAGE 6: NUMERICAL RANGE AND OUTLIER CHECK
# ==========================================

# Select numerical columns for analysis
numeric_columns = [
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph",
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2"
]

# Display minimum, maximum, mean and median
print("Numerical variable summary:")

display(
    df[numeric_columns].agg(["min", "max", "mean", "median"]).T
)

# Calculate outlier boundaries using the IQR method
outlier_summary = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Variable": column,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count
    })

# Display outlier summary
outlier_df = pd.DataFrame(outlier_summary)

print("\nPotential outliers using the IQR method:")
display(outlier_df.round(2))
```

```python
# ==========================================
# STAGE 7: INVESTIGATE EXTREME AQI VALUES
# ==========================================

# Basic AQI statistics
print("AQI statistics:")
display(df["aqi_index"].describe())

# Count AQI values above 500
print("\nAQI values above 500:")
print((df["aqi_index"] > 500).sum())

# Percentage of AQI values above 500
print("\nPercentage of AQI values above 500:")
print(round((df["aqi_index"] > 500).mean() * 100, 2), "%")

# Display the 10 highest AQI observations
print("\nTop 10 highest AQI observations:")

display(
    df.nlargest(10, "aqi_index")[
        ["date_ist", "time_ist", "location", "aqi_index",
         "pm2_5", "pm10", "co", "no2"]
    ]
)

# Maximum AQI for each location
print("\nMaximum AQI by location:")

display(
    df.groupby("location")["aqi_index"]
    .max()
    .sort_values(ascending=False)
)
```

```python
# ==========================================
# STAGE 8: CHECK REPEATED POLLUTION PATTERNS
# ==========================================

from itertools import combinations

# Variables to investigate
pollution_columns = [
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2"
]

# Compare every pair of locations
locations = sorted(df["location"].unique())

comparison_results = []

for loc1, loc2 in combinations(locations, 2):

    data1 = (
        df[df["location"] == loc1]
        .sort_values(["date_ist", "time_ist"])
        .reset_index(drop=True)
    )

    data2 = (
        df[df["location"] == loc2]
        .sort_values(["date_ist", "time_ist"])
        .reset_index(drop=True)
    )

    for column in pollution_columns:

        matching = (data1[column] == data2[column]).mean() * 100

        comparison_results.append({
            "Location 1": loc1,
            "Location 2": loc2,
            "Variable": column,
            "Matching Percentage": round(matching, 2)
        })

# Display comparison results
comparison_df = pd.DataFrame(comparison_results)

display(
    comparison_df.sort_values(
        "Matching Percentage",
        ascending=False
    )
)
```

```python
# ==========================================
# STAGE 9: VERIFY REPEATED DATA PATTERNS
# ==========================================

pollution_columns = [
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2"
]

weather_columns = [
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph"
]

locations = sorted(df["location"].unique())

# Compare pollution and weather records between locations
results = []

for i in range(len(locations)):
    for j in range(i + 1, len(locations)):

        loc1 = locations[i]
        loc2 = locations[j]

        data1 = (
            df[df["location"] == loc1]
            .sort_values(["date_ist", "time_ist"])
            .reset_index(drop=True)
        )

        data2 = (
            df[df["location"] == loc2]
            .sort_values(["date_ist", "time_ist"])
            .reset_index(drop=True)
        )

        pollution_match = (
            data1[pollution_columns] ==
            data2[pollution_columns]
        ).all(axis=1).mean() * 100

        weather_match = (
            data1[weather_columns] ==
            data2[weather_columns]
        ).all(axis=1).mean() * 100

        results.append({
            "Location 1": loc1,
            "Location 2": loc2,
            "Pollution Match (%)": round(pollution_match, 2),
            "Weather Match (%)": round(weather_match, 2)
        })

# Display results
verification_df = pd.DataFrame(results)

display(
    verification_df.sort_values(
        "Pollution Match (%)",
        ascending=False
    )
)
```

```python
# ==========================================
# STAGE 10: CREATE CLEAN WORKING DATASET
# ==========================================

# Create a separate copy of the original dataset
df_clean = df.copy()

# Remove columns not required for our daily analysis
columns_to_remove = [
    "time_ist",
    "lat",
    "lon",
    "condition_text",
    "description"
]

df_clean = df_clean.drop(columns=columns_to_remove)

# Rename date column for easier analysis
df_clean = df_clean.rename(
    columns={"date_ist": "date"}
)

# Arrange columns in a logical order
df_clean = df_clean[
    [
        "date",
        "location",
        "aqi_index",
        "pm2_5",
        "pm10",
        "co",
        "no2",
        "temp_c",
        "humidity",
        "pressure_mb",
        "windspeed_kph"
    ]
]

# Display the cleaned dataset
print("Cleaned dataset preview:")
display(df_clean.head())

# Check dimensions
print("\nCleaned dataset shape:")
print(df_clean.shape)

# Check remaining columns
print("\nRemaining columns:")
print(df_clean.columns.tolist())

# Confirm missing values
print("\nMissing values:")
display(df_clean.isnull().sum())
```

```python
# ==========================================
# STAGE 11: DAILY DATA AGGREGATION
# ==========================================

# Identify numerical columns to aggregate
daily_columns = [
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2",
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph"
]

# Calculate daily averages for each location
df_daily = (
    df_clean
    .groupby(["date", "location"])[daily_columns]
    .mean()
    .reset_index()
)

# Sort records by location and date
df_daily = df_daily.sort_values(
    ["location", "date"]
).reset_index(drop=True)

# Display the daily dataset
print("Daily dataset preview:")
display(df_daily.head(10))

# Check dataset dimensions
print("\nDaily dataset shape:")
print(df_daily.shape)

# Check records per location
print("\nDaily records per location:")
display(df_daily.groupby("location").size())

# Check missing values
print("\nMissing values:")
display(df_daily.isnull().sum())
```

```python
# ==========================================
# STAGE 12: DESCRIPTIVE STATISTICS
# ==========================================

# Select variables for statistical analysis
analysis_columns = [
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2",
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph"
]

# Calculate descriptive statistics
descriptive_stats = df_daily[analysis_columns].describe().T

# Add median and variance
descriptive_stats["median"] = df_daily[analysis_columns].median()
descriptive_stats["variance"] = df_daily[analysis_columns].var()

# Round results for readability
descriptive_stats = descriptive_stats.round(2)

# Display results
print("Descriptive statistics for daily data:")
display(descriptive_stats)

# Calculate average AQI for each location
print("\nAverage daily AQI by location:")

location_aqi = (
    df_daily.groupby("location")["aqi_index"]
    .mean()
    .sort_values(ascending=False)
)

display(location_aqi.round(2))
```

```python
# ==========================================
# STAGE 13: MONTHLY AQI ANALYSIS
# ==========================================

# Extract month from date
df_daily["month"] = df_daily["date"].dt.month

# Calculate monthly average AQI for each location
monthly_aqi = (
    df_daily.groupby(["month", "location"])["aqi_index"]
    .mean()
    .reset_index()
)

# Create month names
monthly_aqi["month_name"] = pd.to_datetime(
    monthly_aqi["month"],
    format="%m"
).dt.strftime("%B")

# Display monthly AQI
print("Monthly average AQI by location:")
display(monthly_aqi.round(2))

# Calculate overall monthly average AQI
overall_monthly_aqi = (
    df_daily.groupby("month")["aqi_index"]
    .mean()
    .reset_index()
)

overall_monthly_aqi["month_name"] = pd.to_datetime(
    overall_monthly_aqi["month"],
    format="%m"
).dt.strftime("%B")

print("\nOverall monthly average AQI:")
display(overall_monthly_aqi.round(2))
```

```python
# ==========================================
# STAGE 14: MONTHLY AQI TREND VISUALISATION
# ==========================================

# Create a figure
plt.figure(figsize=(13, 6))

# Plot monthly AQI for each location
sns.lineplot(
    data=monthly_aqi,
    x="month",
    y="aqi_index",
    hue="location",
    marker="o",
    linewidth=2
)

# Add chart title and labels
plt.title(
    "Monthly Average AQI Trends Across Six Delhi Locations - 2025",
    fontsize=14
)

plt.xlabel("Month")
plt.ylabel("Average Daily AQI")

# Set month labels
plt.xticks(
    range(1, 13),
    ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
     "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
)

plt.legend(title="Location", bbox_to_anchor=(1.02, 1))
plt.tight_layout()

plt.show()
```

```python
# ==========================================
# STAGE 15: POLLUTANT COMPARISON BY LOCATION
# ==========================================

# Select pollution variables
pollutants = ["pm2_5", "pm10", "co", "no2"]

# Calculate average pollutant levels by location
location_pollutants = (
    df_daily.groupby("location")[pollutants]
    .mean()
    .round(2)
)

# Display the results
print("Average pollutant concentrations by location:")
display(location_pollutants)

# Identify the location with the highest average for each pollutant
print("\nLocation with highest average pollutant level:")

for pollutant in pollutants:

    highest_location = location_pollutants[pollutant].idxmax()
    highest_value = location_pollutants[pollutant].max()

    print(
        f"{pollutant}: {highest_location} "
        f"({highest_value:.2f})"
    )
```

```python
# ==========================================
# STAGE 16: POLLUTANT COMPARISON VISUALISATION
# ==========================================

# Create a 2x2 chart layout
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Flatten axes for easier iteration
axes = axes.flatten()

# Plot each pollutant
for i, pollutant in enumerate(pollutants):

    sns.barplot(
        data=location_pollutants.reset_index(),
        x="location",
        y=pollutant,
        ax=axes[i]
    )

    axes[i].set_title(
        f"Average {pollutant.upper()} by Location"
    )

    axes[i].set_xlabel("Location")
    axes[i].set_ylabel(f"Average {pollutant.upper()}")

    axes[i].tick_params(axis="x", rotation=35)

# Adjust spacing
plt.suptitle(
    "Average Pollutant Concentrations Across Delhi Locations - 2025",
    fontsize=15
)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STAGE 17: AQI AND POLLUTANT CORRELATION
# ==========================================

# Select AQI and pollutant variables
correlation_columns = [
    "aqi_index",
    "pm2_5",
    "pm10",
    "co",
    "no2"
]

# Calculate Pearson correlation
correlation_matrix = df_daily[
    correlation_columns
].corr(method="pearson")

# Display correlation values
print("Correlation matrix:")
display(correlation_matrix.round(3))

# Visualise the correlation matrix
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between AQI and Air Pollutants - 2025")
plt.tight_layout()
plt.show()

# Rank pollutant correlations with AQI
print("\nCorrelation of pollutants with AQI:")

aqi_correlations = (
    correlation_matrix["aqi_index"]
    .drop("aqi_index")
    .sort_values(ascending=False)
)

display(aqi_correlations)
```

```python
# ==========================================
# STAGE 18: AQI AND WEATHER CORRELATION
# ==========================================

# Select AQI and weather variables
weather_columns = [
    "aqi_index",
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph"
]

# Calculate Pearson correlation matrix
weather_correlation = df_daily[
    weather_columns
].corr(method="pearson")

# Display correlation matrix
print("AQI and weather correlation matrix:")
display(weather_correlation.round(3))

# Visualise correlation matrix
plt.figure(figsize=(9, 6))

sns.heatmap(
    weather_correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between AQI and Weather Conditions - 2025")
plt.tight_layout()
plt.show()

# Rank weather correlations with AQI
print("\nCorrelation of weather variables with AQI:")

aqi_weather_correlations = (
    weather_correlation["aqi_index"]
    .drop("aqi_index")
    .sort_values(ascending=False)
)

display(aqi_weather_correlations)
```

```python
# ==========================================
# STAGE 19: AQI AND WEATHER SCATTERPLOTS
# ==========================================

# Weather variables to compare with AQI
weather_variables = [
    "temp_c",
    "humidity",
    "pressure_mb",
    "windspeed_kph"
]

# Create a 2x2 chart layout
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes = axes.flatten()

# Generate scatterplots with regression lines
for i, variable in enumerate(weather_variables):

    sns.regplot(
        data=df_daily,
        x=variable,
        y="aqi_index",
        ax=axes[i],
        scatter_kws={"alpha": 0.4},
        line_kws={"color": "red"}
    )

    axes[i].set_title(
        f"AQI vs {variable}"
    )

    axes[i].set_xlabel(variable)
    axes[i].set_ylabel("Daily Mean AQI")

# Add overall title
plt.suptitle(
    "Relationship Between AQI and Weather Conditions - 2025",
    fontsize=15
)

plt.tight_layout()
plt.show()
```
